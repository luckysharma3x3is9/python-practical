```python
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vikram", "Anjali", "Rohan", "Neha", "Karan", "Pooja"],
    "Age": [25, 34, 29, 41, 38, 27, 45, 31, 36, 28],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [1500, 2300, np.nan, 3100, 1800, 2700, 3500, np.nan, 2200, 1900]
})

# First 5 rows
print(customers.head())

# Summary statistics
print(customers.describe())

# Check missing values
print(customers.isnull().sum())

# Handle missing values (fill with mean)
customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(customers["Purchase_Amount"].mean())
print(customers.isnull().sum())
```

```python
# Customers with Age > 30
print(customers[customers["Age"] > 30])

# Total Purchase Amount for male and female separately
print(customers.groupby("Gender")["Purchase_Amount"].sum())

# Only Customer_ID, Name, Purchase_Amount
print(customers[["Customer_ID", "Name", "Purchase_Amount"]])
```

```python
# New column with 10% discount
customers["Discounted_Price"] = customers["Purchase_Amount"] * 0.90
print(customers)

# Average Purchase_Amount by Gender
print(customers.groupby("Gender")["Purchase_Amount"].mean())

# Sort by Purchase_Amount, descending
print(customers.sort_values(by="Purchase_Amount", ascending=False))
```
