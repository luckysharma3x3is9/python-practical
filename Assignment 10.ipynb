```python
%pip install seaborn matplotlib pandas

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load Iris dataset
df = sns.load_dataset('iris')

# Display first five rows
print(df.head())
```

```python
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title("Scatter Plot of Sepal Length vs Petal Length")
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x='petal_length',
    hue='species',
    fill=True,
    common_norm=False
)

plt.title("Density Plot of Petal Length")
plt.show()
```

```python
sns.violinplot(
    data=df,
    x='species',
    y='petal_width'
)

plt.title("Violin Plot of Petal Width by Species")
plt.show()
```

```python
correlation = df.select_dtypes(include='number').corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
sns.jointplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    kind='scatter'
)

plt.show()
```

```python
sns.pairplot(
    df,
    hue='species',
    diag_kind='kde'
)

plt.show()
```
