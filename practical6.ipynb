```python
class Person:

    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):

    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


Emp_details = Emp("Mayank", 103)

# Calling parent class function
Emp_details.Display()

# Calling child class function
Emp_details.Print()
```

```python
def anagram_check(str1, str2):

    if sorted(str1.lower()) == sorted(str2.lower()):
        return True
    else:
        return False


word1 = input("\nEnter first word: ")
word2 = input("Enter second word: ")

if anagram_check(word1, word2):
    print("The words are Anagrams")
else:
    print("The words are not Anagrams")
```

```python
def is_armstrong(number):

    digits = str(number)
    power = len(digits)

    total = 0

    for digit in digits:
        total += int(digit) ** power

    return total == number


number = int(input("\nEnter a number to check Armstrong: "))

if is_armstrong(number):
    print(number, "is an Armstrong Number")
else:
    print(number, "is not an Armstrong Number")
```

```python
class Vehicle:

    color = "white"

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.seating_capacity = 0

    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)


# Creating first vehicle
car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

# Creating second vehicle
car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)

print("\nVehicle 1:")
car1.display_properties()

print("\nVehicle 2:")
car2.display_properties()
```
